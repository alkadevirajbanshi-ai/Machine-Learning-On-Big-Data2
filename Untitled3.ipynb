```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, count, mean, when, isnull

# Connect to Google Drive
from google.colab import drive
drive.mount
```

```python
spark = SparkSession.builder \
                    .appName("Tutorial_DF") \
                    .master("local[*]") \
                    .config("spark.executor.memory", "4g") \
                    .config("spark.driver.memory", "2g") \
                    .config("spark.executor.cores", "2") \
                    .config("spark.sql.inMemoryColumnarStorage.compressed", "true") \
                    .getOrCreate()
```

```python
# Load CSV into DataFrame
file_path = "/content/Airline_Delay_Cause.csv"
df = spark.read.csv(file_path, header=True, inferSchema=True)
```

```python
# Show the first 5 rows
print("Showing first 5 rows:")
df.show(5)

# Print schema of the DataFrame
print("Schema of DataFrame:")
df.printSchema(1)
```

```python
import matplotlib.pyplot as plt

# Calculate the delay ratio
df_delay_ratio = df.withColumn("delay_ratio", col("arr_del15") / col("arr_flights"))

# Group by year and month and calculate the average delay ratio
df_time_series = df_delay_ratio.groupBy("year", "month").agg(
    mean("delay_ratio").alias("average_delay_ratio")
).orderBy("year", "month")

# Convert to Pandas DataFrame for plotting
pandas_df_time_series = df_time_series.toPandas()

# Create a combined 'year_month' column for plotting
pandas_df_time_series['year_month'] = pandas_df_time_series['year'].astype(str) + '-' + pandas_df_time_series['month'].astype(str).str.zfill(2)

# Plotting the line chart
plt.figure(figsize=(15, 7))
plt.plot(pandas_df_time_series['year_month'], pandas_df_time_series['average_delay_ratio'], marker='o', linestyle='-')
plt.title('Overall Flight Delay Ratio Over Time')
plt.xlabel('Year-Month')
plt.ylabel('Average Delay Ratio (arr_del15 / arr_flights)')
plt.grid(True)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
from pyspark.sql.functions import desc

# Calculate the average delay rate per carrier
df_carrier_delays = df.groupBy("carrier_name").agg(
    (mean(col("arr_del15")) / mean(col("arr_flights"))).alias("average_delay_rate")
).orderBy(desc("average_delay_rate"))

# Show the top 10 carriers by delay rate
print("Top 10 Carriers by Average Delay Rate:")
df_carrier_delays.show(10)
```

```python
# Convert to Pandas DataFrame for plotting
pandas_df_carrier_delays = df_carrier_delays.toPandas()

# Plotting the bar chart
plt.figure(figsize=(15, 7))
plt.bar(pandas_df_carrier_delays['carrier_name'], pandas_df_carrier_delays['average_delay_rate'], color='skyblue')
plt.xlabel('Carrier Name')
plt.ylabel('Average Delay Rate')
plt.title('Average Flight Delay Rate per Carrier')
plt.xticks(rotation=90, ha='right')
plt.tight_layout()
plt.show()
```

```python
from pyspark.sql.functions import sum

# Calculate the total carrier-caused delays per airline
df_carrier_ct = df.groupBy("carrier_name").agg(
    sum("carrier_ct").alias("total_carrier_caused_delays")
).orderBy(desc("total_carrier_caused_delays"))

# Show the top 10 carriers by carrier-caused delays
print("Top 10 Carriers by Total Carrier-Caused Delays:")
df_carrier_ct.show(10)
```

```python
# Convert to Pandas DataFrame for plotting
pandas_df_carrier_ct = df_carrier_ct.toPandas()

# Plotting the bar chart
plt.figure(figsize=(15, 7))
plt.bar(pandas_df_carrier_ct['carrier_name'], pandas_df_carrier_ct['total_carrier_caused_delays'], color='lightcoral')
plt.xlabel('Carrier Name')
plt.ylabel('Total Carrier-Caused Delays')
plt.title('Total Carrier-Caused Delays per Airline')
plt.xticks(rotation=90, ha='right')
plt.tight_layout()
plt.show()
```

```python
# Calculate the average delay rate per airport
df_airport_delays = df.groupBy("airport_name").agg(
    (mean(col("arr_del15")) / mean(col("arr_flights"))).alias("average_delay_rate")
).orderBy(desc("average_delay_rate"))

# Show the top 10 airports by delay rate
print("Top 10 Airports by Average Delay Rate:")
df_airport_delays.show(10)
```

```python
# Convert to Pandas DataFrame for plotting
pandas_df_airport_delays = df_airport_delays.toPandas()

# Plotting the bar chart
plt.figure(figsize=(15, 7))
plt.bar(pandas_df_airport_delays['airport_name'], pandas_df_airport_delays['average_delay_rate'], color='lightgreen')
plt.xlabel('Airport Name')
plt.ylabel('Average Delay Rate')
plt.title('Average Flight Delay Rate per Airport')
plt.xticks(rotation=90, ha='right')
plt.tight_layout()
plt.show()
```

```python
# Group by year and sum the delay causes
df_delay_causes_by_year = df.groupBy("year").agg(
    sum("carrier_ct").alias("carrier_delays"),
    sum("weather_ct").alias("weather_delays"),
    sum("nas_ct").alias("nas_delays"),
    sum("security_ct").alias("security_delays"),
    sum("late_aircraft_ct").alias("late_aircraft_delays")
).orderBy("year")

# Convert to Pandas DataFrame
pandas_df_delay_causes = df_delay_causes_by_year.toPandas()

# Calculate total delays for percentage calculation
pandas_df_delay_causes['total_delays'] = pandas_df_delay_causes[['carrier_delays', 'weather_delays', 'nas_delays', 'security_delays', 'late_aircraft_delays']].sum(axis=1)

# Calculate percentages
for col_name in ['carrier_delays', 'weather_delays', 'nas_delays', 'security_delays', 'late_aircraft_delays']:
    pandas_df_delay_causes[col_name + '_pct'] = (pandas_df_delay_causes[col_name] / pandas_df_delay_causes['total_delays']) * 100

# Display the percentages (first few rows)
print("Delay Cause Distribution by Year (Percentages):")
display(pandas_df_delay_causes.head())
```

```python
import numpy as np

# Prepare data for stacked bar chart
plot_data = pandas_df_delay_causes[['year', 'carrier_delays_pct', 'weather_delays_pct', 'nas_delays_pct', 'security_delays_pct', 'late_aircraft_delays_pct']]
plot_data = plot_data.set_index('year')

# Plotting the stacked bar chart
plt.figure(figsize=(16, 8))
plot_data.plot(kind='bar', stacked=True, colormap='viridis', figsize=(15, 7))
plt.title('Distribution of Delay Causes by Year (Percentage)')
plt.xlabel('Year')
plt.ylabel('Percentage of Delays')
plt.xticks(rotation=45)
plt.legend(title='Delay Cause', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
from pyspark.sql.functions import sum

# Group by month and sum the delay causes
df_delay_causes_by_month = df.groupBy("month").agg(
    sum("carrier_ct").alias("carrier_delays"),
    sum("weather_ct").alias("weather_delays"),
    sum("nas_ct").alias("nas_delays"),
    sum("security_ct").alias("security_delays"),
    sum("late_aircraft_ct").alias("late_aircraft_delays")
).orderBy("month")

# Convert to Pandas DataFrame for plotting
pandas_df_delay_causes_month = df_delay_causes_by_month.toPandas()

print("Delay Causes by Month:")
display(pandas_df_delay_causes_month.head())
```

```python
# Plotting the multi-line chart
plt.figure(figsize=(15, 8))

plt.plot(pandas_df_delay_causes_month['month'], pandas_df_delay_causes_month['carrier_delays'], label='Carrier Delays', marker='o')
plt.plot(pandas_df_delay_causes_month['month'], pandas_df_delay_causes_month['weather_delays'], label='Weather Delays', marker='o')
plt.plot(pandas_df_delay_causes_month['month'], pandas_df_delay_causes_month['nas_delays'], label='NAS Delays', marker='o')
plt.plot(pandas_df_delay_causes_month['month'], pandas_df_delay_causes_month['security_delays'], label='Security Delays', marker='o')
plt.plot(pandas_df_delay_causes_month['month'], pandas_df_delay_causes_month['late_aircraft_delays'], label='Late Aircraft Delays', marker='o')

plt.title('Distribution of Delay Causes by Month')
plt.xlabel('Month')
plt.ylabel('Total Count of Delays')
plt.xticks(range(1, 13), ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'])
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend(title='Delay Cause')
plt.tight_layout()
plt.show()
```

```python
# Calculate total delayed flights and total flights per month
df_monthly_delays = df.groupBy("month").agg(
    sum("arr_del15").alias("total_delayed_flights"),
    sum("arr_flights").alias("total_flights")
).orderBy("month")

# Calculate the delay percentage per month
df_monthly_delays = df_monthly_delays.withColumn(
    "delay_percentage",
    (col("total_delayed_flights") / col("total_flights")) * 100
)

# Convert to Pandas DataFrame for plotting
pandas_df_monthly_delays = df_monthly_delays.toPandas()

print("Monthly Delay Percentages:")
display(pandas_df_monthly_delays.head())
```

```python
# Plotting the line chart for monthly delay percentages
plt.figure(figsize=(15, 7))
plt.plot(pandas_df_monthly_delays['month'], pandas_df_monthly_delays['delay_percentage'], marker='o', linestyle='-', color='purple')

plt.title('Overall Flight Delay Percentage by Month')
plt.xlabel('Month')
plt.ylabel('Delay Percentage (%)')
plt.xticks(range(1, 13), ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'])
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
